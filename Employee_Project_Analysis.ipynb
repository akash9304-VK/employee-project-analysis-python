# TASK 2 : Fill missing values using running average

```python
# Load project data
df_proj = pd.read_csv('project.csv')
```

```python
df_proj                              # Displaying the DataFrame
```

```python
for i in range(len(df_proj)):                        # using for loop 
    if pd.isna(df_proj.loc[i, 'Cost']):              # applying if conditions to check null 
        
        prev_values = df_proj.iloc[:i]['Cost']       # all the values till null values
        
        avg = prev_values.mean()                # average values of number till null
        
        df_proj.loc[i, 'Cost'] = avg                # adding the calculated averge in place of null 
print(df_proj)                                       # printing the dataframe
```

  #  TASK 3 : Split Name Column and Remove Original Column

```python
# Load Employee data
df_emp = pd.read_csv('employee.csv')
```

```python
df_emp                                  # Displaying the DataFrame
```

#  TASK 4 : Merging employee, project, and seniority datasets into a single dataframe

```python
# Load Seniority data
df_slevel = pd.read_csv('Seniority_level.csv')
```

```python
Final_df = pd.merge(df_proj,df_emp, on = 'ID')       # Merging  Project and Employee dataframes on 'ID'
```

```python
Final_df                                    # Displaying the merged dataframe
```

```python
Final_df = pd.merge(Final_df,df_slevel, on = 'ID')         # Merging the result seniority dataframe on 'ID'
```

```python
Final_df                                    # Displaying the final merged dataframe
```

# Task 5 : Create Bonus Column

```python
Final_df['Bonus'] = (Final_df['Status'] == 'Finished') * Final_df['Cost'] * 0.05  # Creating 'Bonus' column: 5% of Cost for 'Finished' projects, else 0
```

```python
Final_df                        # Displaying the DataFrame
```

# Task 6 : Demote and Remove Employees

```python
Final_df.loc[Final_df['Status'] == 'Failed', 'Designation Level'] -= 1             # Decrease designation level by 1 for employees with 'Failed' status column
```

```python
Final_df = Final_df[Final_df['Designation Level'] <= 4]                        # Keep only rows where Designation Level is less than or equal to 4
```

```python
Final_df                                 # Displaying the DataFrame
```

# Task 7 : Add Mr & Mrs to First Name and Drop Gender Column

```python
# Adding 'Mr' for males and 'Mrs' for females in the First Name column
Final_df.loc[Final_df['Gender'] == 'M', 'First Name'] = 'Mr' + ' ' + Final_df['First Name']
Final_df.loc[Final_df['Gender'] == 'F', 'First Name'] = 'Mrs' + ' ' + Final_df['First Name']
```

```python
Final_df                                 # Displaying the DataFrame
```

```python
Final_df.drop('Gender', axis = 1 , inplace = True)                       # using drop function to drop 'Gender' column from the dataframe
```

```python
Final_df                  # Displaying the final dataframe
```

# Task 8 : Promotion using IF Condition 

```python
for i in range(len(Final_df)) :                         # Using a for loop to iterate through the DataFrame
    if Final_df.loc[i,'Age'] > 29 :                    # Applying IF condition to check Age > 29                    
        Final_df.loc[i,'Designation Level'] += 1       # Increasing Designation Level by 1 if condition is satisfied
        
```

```python
Final_df                          # Displaying the DataFrame
```

# Task 9 : Calculate Total Project Cost per Employee

```python
TotalProjCost = Final_df.groupby(['ID','First Name'])['Cost'].sum().reset_index()   # Creating new DataFrame using groupby and sum aggregation
```

```python
TotalProjCost                         # Displaying the aggregated DataFrame
```

```python
TotalProjCost.rename(columns = {'Cost' : 'Total Cost'}, inplace = True)            # Renaming 'Cost' column to 'Total Cost'
```
